## Introduction to roboquant
[roboquant](https://python.roboquant.org/) is an open-source algorithmic trading platform designed for speed and flexibility. In this tutorial, we will walk through the core components of a backtest:
1. **Environment Setup**: Installing and configuring the library.
2. **Feed**: Getting historical market data.
3. **Strategy**: Defining the logic for buying and selling.
4. **Run**: Executing the backtest and viewing results.
5. **Visualization**: Analyzing the performance.

## 1. Setup and Imports
First, we install the package and then import the library. We also set a visual style for our charts to make them more readable.

```python
# Install roboquant
%pip install --quiet --upgrade roboquant
```

```python
import roboquant as rq

rq.set_light_style()
# uncommon following line if you prefer dark styled charts
# rq.set_dark_style()
```

## 2. The Data Feed
A **Feed** represents a source of market data (like stock prices). Here, we use a built-in feed that fetches data for 10 popular US stocks via Yahoo Finance.

```python
feed = rq.feeds.YahooFeed.us_stocks_10()
```

## 3. Strategy and Execution
We use the built-in `EMACrossover` strategy, which generates signals when a fast Exponential Moving Average crosses a slow one.

```python
strategy = rq.strategies.EMACrossover()
```

## 4. Run
The `run` function connects our feed to the strategy and simulates trading, returning an `Account` object that holds the results.

```python
account = rq.run(feed, strategy)
print(account)
```

## 5. Analyzing Performance
Finally, we visualize the price action of a specific symbol (e.g., JPM) and overlay the trades executed by our strategy to see where it entered and exited positions.

```python
feed.plot("JPM", trades=account.trades);
```
